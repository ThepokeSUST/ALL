# For Index

```python
import pandas as pd
import numpy as np
data = {'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'],
        'Age': [25, 30, 22, 35, 28],
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'],
        'Salary': [50000, 55000, 40000, 70000, 48000]}

df = pd.DataFrame(data,index=[2,3,4,5,6])
print(df)
print()
print(df.index)  # Accessing the index
```

### Custom index

```python
data = {'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'],
        'Age': [25, 30, 22, 35, 28],
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'],
        'Salary': [50000, 55000, 40000, 70000, 48000]}

df = pd.DataFrame(data)
print(df)
print()
new_df_Index=df.set_index('Age')
print(df)
print(new_df_Index,type(new_df_Index))
```

## reset_index

```python
print(df)
new_df=df.reset_index()
print(new_df)
print(new_df.reset_index())
print(new_df_Index.reset_index())
```
