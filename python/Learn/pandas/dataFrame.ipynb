# This is for learning pandas Data Frame

## this code for create dataFrame from list data structure

```python
import pandas as pd
import numpy as np
df=pd.DataFrame()

lst=np.array([12,22,33,44])
df=pd.DataFrame(lst)
print(df)
```

#### this code for create dataFrame from dictionary data stucture
##### Each key in the dictionary represents a column name

```python
tp=(1,2,3)
dct={
    "name":["dipok"],
    "dept": ["Cse"],
     "tp": [tp]
}

df=pd.DataFrame(dct)
print(df)
```

```python
dct={
    "name":["dipok"] * len(tp),
    "dept": ["Cse"]* len(tp),
     "tp": tp
}
df=pd.DataFrame(dct)
print(df)
```

##### use 2d list to create data frame

```python
lst=np.array([[12,4,4],
              [4,5,6],
              [67,7,675]])
print("list \n",lst)

df=pd.DataFrame(lst)
print("Data Frame\n",df)
```
