## this is our first pandas tutorial


```python
import pandas as pd
import numpy as np
df= pd.DataFrame()
print(df)
```

## create a data frame from a list

```python
lst = ["geeks","for","geek"]
df=pd.DataFrame(lst,index=[2,3,4])
print(df)
```

## from numpy Array

```python
# import numpy as np
# import pandas as pd
data = np.array([1,2,3,4])
print(data,data.shape)
print()
df= pd.DataFrame(data,index=[2,3,5,8],columns=['A'])
print(df)
```

```python
data = np.array([[1,2,3,4],[5,6,7,8]])
print(data,data.shape)
print()
df= pd.DataFrame(data)
print(df)
```

## DataFrame expects 1D or 2D-like data (rows and columns), not 3D.

```python
data= np.array([[[1,2,3,4],
                 [3,4,5,5]],
                 [[7,8,9,0],
                  [9,8,7,6]]])
print(data,data.shape)

# df=pd.DataFrame(data)
# print(df)
```

### create dataframe using dictionary

#### DataFrame expects each value in the dictionary to be a list (or array) of the same length.

```python
dct={
    "name":"dipok",
    "age": 21

}
print(dct)
# df=pd.DataFrame(dct)
```

```python
df=pd.DataFrame([dct])
print(df)
```

```python
dct={
    "name": ["dipok","dipu","rifat","susmit","shimul","mamun"],
    "age" : [21,21,23,24,25,34]#all array must be the same length...
}

df=pd.DataFrame(dct)
print(df)
```
