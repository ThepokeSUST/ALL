# pop function

```python
import pandas as pd
data={

    'name':['A','B'],
    'age':[2,3]
}

df=pd.DataFrame(data)
print(df,end='\n\n')
deleted=df.pop('name')
print(df)
print(deleted,type(deleted))
print()
print(df)
```

## Get method


```python
df['name']=deleted
print(df)
df.index=['a','b']

getname=df.get('name')
print(df)
print(getname,type(getname))
print()
get=df.get(['name'])
print(get,type(get))
print('new')
getall=df.get(['age','name','b'])
print()
print(getall)

g=df.get('name')
print(g)
g=df.get('a')
print()
print(g)
```
