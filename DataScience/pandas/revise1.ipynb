## isin

```python
data={
     "name":['dipok','dipu','rifta','susmit','ayesha','sadia'],
     'age': [23,23,45,56,67,78],
     'dept':['cse','swe','eee','ipe','fet','pme']
}

df=pd.DataFrame(data)
# print(df)
print()

d=df['age'].isin([23])
print(d,type(d))
df1=df[d]
print(df1)
print()
d=df[(df['age']>23) & (df['dept']=='cse')]
print(d)

print()
d=df['age'].isin([23,45])
print(d)
print(df[d])
```

## pop()

```python
data={
     "name":['dipok','dipu','rifta','susmit','ayesha','sadia'],
     'age': [23,23,45,56,67,78],
     'dept':['cse','swe','eee','ipe','fet','pme']
}

df=pd.DataFrame(data)
# print(df)
print()

deleted=df.pop('dept')
print(deleted,type(deleted))
print(df)

df['dept']=deleted
# print(df)

# df.pop(['name'])
d=df.drop(['name'],axis=1,inplace=False)
print(df)
print(d)
```

## insert

```python
data={
     "name":['dipok','dipu','rifta','susmit','ayesha','sadia'],
     'age': [23,23,45,56,67,78],
     'dept':['cse','swe','eee','ipe','fet','pme']
}

df=pd.DataFrame(data)
# print(df)
print()
print(df)
df.insert(1,'newcol',[1,2,3,4,5,6])
print(df)

df.insert(1,'newcol1',100)
# print(df)
df.insert(1,'age',30,allow_duplicates=True)
print(df)

# print(df['age'],type(df['age']))
filt=df[df['newcol']>1]
print(filt)
# df.insert(1,'age',30,allow_duplicates=False)
```
