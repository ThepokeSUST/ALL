```python
# importing pandas
import pandas as pd

# creating dataframe
df = pd.DataFrame({'Product': ['Carrots', 'Broccoli', 'Banana', 'Banana',
                               'Beans', 'Orange', 'Broccoli', 'Banana'],
                   'Category': ['Vegetable', 'Vegetable', 'Fruit', 'Fruit',
                                'Vegetable', 'Fruit', 'Vegetable', 'Fruit'],
                   'Quantity': [8, 5, 3, 4, 5, 9, 11, 8],
                   'Amount': [270, 239, 617, 384, 626, 610, 62, 90]})
df
```

## Get the Total Sales of Each Product

```python
pivot=pd.pivot_table(df,index=['Product'],values=['Amount','Quantity'],aggfunc='sum')
print(pivot)
```

# Get the Total Sales of Each Category

```python
pivot=df.pivot_table(index='Category',values='Amount',aggfunc='sum')
print(pivot)
```

## mean min median

```python
pivot=df.pivot_table(index='Category',values='Amount',aggfunc=['sum','mean','median','min'])
print(df)
print()
print(df[df['Category']=='Vegetable'])
print()
print(df[df['Category']!='Vegetable'])
print()
print(pivot)
```
