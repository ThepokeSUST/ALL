# Using List


```python
import  pandas as pd
import numpy as np

data=[1,23,4,5]

df=pd.DataFrame(data)

print(df)
```

# Using Numpy array

```python
data= np.array([[1,2,3,4,5,6,7],
                [8,6,5,4,3,0,6]])
# print(data)
df= pd.DataFrame(data,columns=['A','B','C','D','E','F','G'],index=['i','ii'])
print(df)
```

# From list of Dictionaries

```python
dic={
    "Name":['dipok','dipu','fipok','fipu'],
    'Age':[21,22,20,23],
    'Dept':['Cse','Swe',"Eee",'Cse']
}
df=pd.DataFrame(dic,index=[1,2,3,4])
print(df)
df=pd.DataFrame(dic,index=[1,2,3,4],columns=['Name',1,'Dept'])
print(df)
```

```python
dic={
    "Name":['dipok','dipu','fipok','fipu'],
    'Age':[21,22,20,23],
    'Dept':['Cse','Swe',"Eee",'Cse']
}


df= pd.DataFrame(dic,columns=[100,'Name'],index=[1,2,3,4])
print(df)
```

## from list of Dict


```python
data=[{'b':2},{'c':3},{'d':4}]
print(data,type(data))

print()

df=pd.DataFrame(data)
print(df)
df=pd.DataFrame(data,index=[1,2,3])
print(df)

data=[{'b':2,'c':3},{'d':4,'b':5}]
print(data)
df=pd.DataFrame(data)
print(df)
print()
print()
print(data)
df=pd.DataFrame(data , columns=['1',2])
print(df)
print(type(df),"end")
```

## from Another Dict


```python

data={
    
    'name':["dipok",'dipu'],
    'age':[12,34],
    'dept':['cse','swe']
}
df=pd.DataFrame(data)
print(df)
print()
df2=df[['name']]
print(df2,type(df2))
```

## Using zip Function

```python
Name=['tom','krish','nick','dip','dipok']
Age=[21,32,21,22,23,8]

List_of_Tuples=list(zip(Age,Name))
print(List_of_Tuples,type(List_of_Tuples))

df=pd.DataFrame(List_of_Tuples,columns=['Age','Name'])
print(df)
```
