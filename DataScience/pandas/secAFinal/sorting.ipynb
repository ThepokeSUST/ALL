```python
import pandas as pd
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age': [50, 30, 35, 40],
        'Score': [85, 90, 95, 80]}
df = pd.DataFrame(data)
print(df)
```

```python
sortedDf=df.sort_values(by='Age',ascending=False)
print(sortedDf)
sortedIndex=df.sort_index(ascending=False)
print(sortedIndex)
```
