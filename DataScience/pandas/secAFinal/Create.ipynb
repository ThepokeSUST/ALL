## via list

```python
import pandas as pd

lst=[4,3,2,1]

df=pd.DataFrame(lst)

print(df)

print(df.index,"      ",df.columns)
idx=range(0,4,1)
print(idx,type(idx))
```

```python
print(df)
df.reset_index(inplace=True)
print(df)
df.reset_index(inplace=True)
print(df)
df.rename(columns={'level_0':'level_0old'},inplace=True)
df.reset_index(inplace=True,allow_duplicates=False)
print(df)
print(type(df.columns))
print(df.columns)
print(list(df.columns))

```

## rename and sort index

```python
print(df)
df.drop(['level_0','level_0old'],axis=1,inplace=True)
print(df)
df.rename(index={0:1000,0:333,1000:9999},inplace=True)
print(df)
df.sort_index(inplace=True)
print(df)
```

## sort via column

```python
data={
    'name':['dipok','dipu','fipook','fipu'],
    'age':[21,23,23,14],
    'salary':[3232,23,20,32321]
}
df=pd.DataFrame(data)
print(df)
df.sort_index(inplace=True,ascending=False)
print(df)
print()
df.sort_values(by='name',inplace=True)
print(df)
print()
# df.loc[1,'salary']=df.loc[1,'age']
print(".............")
print(df)
print()
df.sort_values(by=['age','salary'],inplace=True)
print(df)
```

```python
print(df)
df.rename(index={1:'one'},columns={'name':'Name'},inplace=True)
print(df)
df.drop('one',axis=0,inplace=True)
df.drop(['age','salary'],axis=1,inplace=True)
print(df)
```

```python
# Initialize data of lists
data = [{'b': 2, 'c': 3}, {'a': 10, 'b': 20, 'c': 30},{'z':0}]

# Creates pandas DataFrame by passing
# Lists of dictionaries and row index.
df = pd.DataFrame(data, index=['first', 'second',"3rd"])

print(df)
```

```python
# Initialize data to Dicts of series.
d = {'one': pd.Series([10, 20, 30, 40],
                      index=['z', 'b', 'f', 'Z']),
     'two': pd.Series([10, 20, 30, 40],
                      index=['a', 'e', 'c', 'd'])}

# creates Dataframe.
df = pd.DataFrame(d)

print(df)
```

# zip() function

```python
lst1=[1,2,3,4,0,9]
lst2=[5,6,7,8,5]
lst3=[11,22,33,44,55,99]
lst=zip(lst1,lst2,lst3)
print(type(lst))
lst=list(lst)
print(type(lst),lst)

df=pd.DataFrame(lst)
print(df)
```

# list of dictionary

```python
data=[{'a':1,'b':2,'d':5},
      {'b':4,'c':3}]

df=pd.DataFrame(data)
print(df)

df2=pd.DataFrame.from_records(data)
print()
print(df)

df3=pd.DataFrame.from_dict(data)
print()
print(df3)
```
