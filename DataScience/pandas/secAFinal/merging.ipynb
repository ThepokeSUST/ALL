```python
import pandas as pd

data1 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],}

data2 = {'key': ['K', 'K1', 'K2', 'K3'],
         'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
        'Qualification':['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1)

df1 = pd.DataFrame(data2)


print(df, "\n\n", df1)
```

```python
res = pd.merge(df, df1, on='key')

print(res)
```

```python
import pandas as pd

data1 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K1', 'K0', 'K1'],
         'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],}

data2 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K0', 'K0', 'K0'],
         'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
        'Qualification':['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1)

df1 = pd.DataFrame(data2)


print(df, "\n\n", df1)
```

```python
mdf=pd.merge(df,df1,on=['key','key1'])
print(mdf)
```

```python
mdf=pd.merge(df,df1,on=['key','key1'],how='inner')
print(mdf)
print()
mdf=pd.merge(df,df1,on=['key','key1'],how='outer')
print(mdf)
```

```python
print(df,'\n',df1)
mdf=pd.merge(df,df1,on=['key','key1'],how='left')
print(mdf)
```

```python
print(df,'\n',df1)
mdf=pd.merge(df,df1,on=['key','key1'],how='right')
print(mdf)
```

# join

```python
import pandas as pd


data1 = {'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32]}

data2 = {'Address':['Allahabad', 'Kannuaj', 'Allahabad', 'Kannuaj'],
        'Qualification':['MCA', 'Phd', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1,index=['K0', 'K1', 'K2', 'K3'])

df1 = pd.DataFrame(data2, index=['K1', 'K2', 'K3', 'K4'])


print(df, "\n\n", df1)
```

```python
print(df.join(df1))
print()
print(df1.join(df))
```
