```python
import pandas as pd


data1 = {'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32]}

data2 = {'Address':['Allahabad', 'Kannuaj', 'Allahabad', 'Kannuaj'],
        'Qualification':['MCA', 'Phd', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1,index=['K0', 'K1', 'K2', 'K3'])

df1 = pd.DataFrame(data2, index=['K1', 'K2', 'K3', 'K4'])


print(df, "\n\n", df1)
```

```python
print(df.join(df1,how='outer'))
```

```python
df.insert(len(df.columns),'K',['KK','K4','K3','K1'])
```

```python

print(df, "\n\n", df1)
print(df.join(df1,on=['K']))
print()
print(df.join(df1,how='inner',on=['K']))
print(df.join(df1,how='outer',on=['K']))


```
