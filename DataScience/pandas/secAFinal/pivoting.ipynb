```python
# importing pandas
import pandas as pd

# creating dataframe
df = pd.DataFrame({'Product': ['Carrots', 'Broccoli', 'Banana', 'Banana',
                               'Beans', 'Orange', 'Broccoli', 'Banana'],
                   'Category': ['Vegetable', 'Vegetable', 'Fruit', 'Fruit',
                                'Vegetable', 'Fruit', 'Vegetable', 'Fruit'],
                   'Quantity': [8, 5, 3, 4, 5, 9, 11, 8],
                   'Amount': [270, 239, 617, 384, 626, 610, 62, 90]})
df
```

```python
pivotTable=df.pivot_table(index='Product',values='Quantity',aggfunc=['sum','min','mean'])
print(pivotTable)
```

```python
pt=df.pivot_table(index=['Product'],values=['Quantity'],aggfunc=['sum'])
print(pt,type(pt))
```

```python
pt1=df.pivot_table(index=['Product','Category'],values=['Quantity'],aggfunc=['mean','sum','min'])
print(pt1)
```

```python
pt1=df.pivot_table(index=['Category','Product'],values=['Quantity'],aggfunc=['mean','sum','min'])
print(pt1)
```
