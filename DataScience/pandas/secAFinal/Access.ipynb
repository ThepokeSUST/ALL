```python
import pandas as pd

data = {'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'],
        'Age': [25, 30, 22, 35, 28],
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'],
        'Salary': [50000, 55000, 40000, 70000, 48000]}

df = pd.DataFrame(data)
print(df)
```

# loc[]

```python
print(df)
print()
print(len(df))
row=df.loc[3]
print(row)
print()
rows=df.loc[0:len(df)]
print(rows)
print()
print(df.loc[0:10])

```

```python
rowCol=df.loc[0:10,'Name':]
print()
print(rowCol)
```

```python
rowCol=df.loc[0:10,['Name','Salary','Age','Gender']]
print()
print(rowCol)
```

# access column

```python
age_column=df['Age']
print(age_column,type(age_column))
```

```python
columns=df[['Name','Age']]
print(columns,type(columns))
```

# iloc[]

```python
print(df)
print()
row=df.iloc[0]
print(row,type(row))
```

# Query 

```python
print(df)
newdata= df.query('Age>22 & Gender=="Male"')
print(newdata,type(newdata))
```

# pop function used to remove specific colunm

```python
print(df)
poped=df.pop(item='Age')
print()
print(df)
print()
print(poped)
# df.insert(1,poped)
print(df)
```

# get function

```python
print(df)
geted= df.get('Name')
print(geted,type(geted))
print()
geted=df.get(['Name','Salary'],[1,2,3,4])
print(geted,type(geted))

notget=df.get('Nothing')
print(notget,type(notget))

notget=df.get('Nothing',[2,3,4])
print(notget,type(notget))
```

# insert Function

```python
print(df)
poped=df.pop('Gender')
print(df)
print()
print(poped)
df.insert(1,'Gender',poped)
print(df)
```

```python
print(df)
isin=df.isin(['Name'])
print(isin)
print()
isin=df.isin(['Male','John','Fema'])
print(isin,type(isin))
newdf=df[isin]
print(newdf)
```

```python
print(newdf)
print(df)
isin=df['Gender'].isin(['Male',"ok"])
print(isin,type(isin))
newdf=df[isin]
print(newdf)
```

# filter1 & filter2

```python
print(df)

filter1=df['Name'].isin(['John'])
print(filter1)
print()
filter2=df['Gender'].isin(['Male'])
print(filter2)
print()
filter3=filter1&filter2
print(filter3,type(filter3))

print(df[filter3],filter3.name)
```

# series logical operation

```python
series1= pd.Series([True,False,True],index=[4,3,2])
print(series1)
series2= pd.Series([True,True,True])
print(series2)

series3=series1&series2
print(series3)
print()
print(series1|series2)
```
