```python
import pandas as pd

data1 = {'Name': ['Jai', 'Princi', 'Gaurav', 'Anuj'],
         'Age': [27, 24, 22, 32],
         'Address': ['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
         'Qualification': ['Msc', 'MA', 'MCA', 'Phd']}

data2 = {'Name': ['Abhi', 'Ayushi', 'Dhiraj', 'Hitesh'],
         'Age': [17, 14, 12, 52],
         'Address': ['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
         'Qualification': ['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1, index=[0, 1, 2, 3])

df1 = pd.DataFrame(data2, index=[4, 5, 6, 7])

print(df, "\n\n", df1)
```

```python
Frames=[df,df1]
df2=pd.concat(Frames,axis=0)
print(df2)

print()
print(pd.concat(Frames,axis=1))
```

```python
df1.rename(index={4:0},inplace=True)
df2=pd.concat(Frames,axis=1)
print(df2)
```

```python
print(pd.concat(Frames,axis=0,ignore_index=True))
print()
print(pd.concat(Frames,axis=1,ignore_index=True))
```

# join=

```python
data1 = {'Name': ['Jai', 'Princi', 'Gaurav', 'Anuj'],
         'Age': [27, 24, 22, 32],
         'Address': ['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
         'Qualification': ['Msc', 'MA', 'MCA', 'Phd']}

data2 = {'Name': ['Abhi', 'Ayushi', 'Dhiraj', 'Hitesh'],
         'Age': [17, 14, 12, 52],
         'Address': ['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
         'Qualification': ['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1, index=[0, 1, 2, 3])

df1 = pd.DataFrame(data2, index=[0, 5, 3, 7])

print(df, "\n\n", df1)
Frames=[df,df1]
print(pd.concat(Frames,join='inner',axis=1))
```

```python
data1={
    'Name':['A','B'],
    'Age':[23,33]
}
data2={
    'Name':['A','B'],
    'Age':[23,33],
    'Dept':['cse','swe']
}

df=pd.DataFrame(data1)
df1=pd.DataFrame(data2)
print(df,"\n",df1)

Frames=[df,df1]
print(pd.concat(Frames,join='inner'))
print(pd.concat(Frames,axis=1))

```

# Sort=

```python
data1={
    'Name':['A','B'],
    'Age':[23,33]
}
data2={
    'Name':['A','B'],
    'Age':[23,33],
    'Dept':['cse','swe']
}

df=pd.DataFrame(data1)
df1=pd.DataFrame(data2)
print(df,"\n",df1)

Frames=[df,df1]
print(pd.concat(Frames,join='outer',sort=True))
print(pd.concat(Frames,axis=1,sort=True))
```

```python
data1={
    'Name':['A','B'],
    'Age':[23,33]
}
data2={
    'Name':['A','B'],
    'Age':[23,33],
    'Dept':['cse','swe']
}

df=pd.DataFrame(data1)
df1=pd.DataFrame(data2)
print(df,"\n",df1)

print(pd.concat([df,df1],axis=1,ignore_index=True))
```

```python
data1={
    'Name':['A','B'],
    'Age':[23,33]
}
data2={
    'Name':['A','B'],
    'Age':[23,33],
    'Dept':['cse','swe']
}

df=pd.DataFrame(data1)
df1=pd.DataFrame(data2)
print(df,"\n",df1)

print(pd.concat([df,df1],axis=0,keys=['A','B']))
print(pd.concat([df,df1],axis=1,keys=['A','B']))
```

# DataFrame and series

```python
data1={
    'Name':['A','B'],
    'Age':[23,33]
}
df=pd.DataFrame(data1)
print(df)
s1=pd.Series([2,3,4])
print()
print(s1)

print(pd.concat([df,s1],join='outer'))
print()
print(pd.concat([df,s1],join='inner',axis=1))
print()
cs=pd.concat([s1,s1],join='inner')
print(cs,type(cs))
```

```python
cs=pd.concat([s1,s1,s1],join='inner',axis=1,ignore_index=True)
print(cs,type(cs))
```
