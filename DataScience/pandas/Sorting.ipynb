```python
import pandas as pd 


data={
    'name':['dipok','dipu','shipu','mipu','salek','malek'],
    'age':[21,23,22,21,24,32]
}

df=pd.DataFrame(data)

print(df)

d=df.sort_values(by='age')
print(df,d,sep='\n\n')

d=d.sort_index()
display(d)
```

```python
df=df.sort_values(by='age',ascending=False)
print(df)
```

```python
data1={
    "name":['A','B','C'],
    'salary':[200,233,232]
}
data2={
    'name':['C','D','E'],
    'age':[23,20,22]
}

df12=pd.concat([pd.DataFrame(data1),pd.DataFrame(data2)])
print(df12)

df12.sort_values(by='salary',inplace=True,ascending=False,na_position='last')
print(df12)
```

## multiple column

```python
data={
    "name":['A','B','C','D'],
    'age':[1,2,2,4],
    'salary':[4,3,2,1]
}
df=pd.DataFrame(data)
print(df)
df.sort_values(by=['salary','age','salary','age'],inplace=True,ascending=False)

print(df)
```
