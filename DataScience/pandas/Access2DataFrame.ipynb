## head mthod


```python
import pandas as pd
data={
    'name':['dipu','dipok','fipok','fipu','thepoke','thepu'],
    'age':[23,21,22,20,22,20],
    'dept':['cse','swe','eee','ipe','civil','archi']
}

df=pd.DataFrame(data)
newdf=df.head()
print(newdf,type(newdf))
print(df.head(10))
```

## tail

```python
print(df.tail())
print()
print(df.tail(10))
```

# at()

```python
print(df)
print(df.at[0,'dept'],type(df.at[0,'dept']))
```

## iat[]

```python
print(df)
print()
print(df.iat[1,2])
```

```python
newdf = df[(df['age'] > 20) & ((df['dept'] == 'cse') | (df['dept'] == 'sew'))]
print(newdf)
```

## Query

```python
data={
    'name':['dipok','dipu','mahir','susmit','dipika','susmita','dipa'],
    'age':[21,20,23,24,20,23,22],
    'dept':['cse','swe','bba','agri','phy','chem','bmb'],
    'gender':['male','male','male','male','female','female','female',]
}
df=pd.DataFrame(data)
print(df)
print()
df2=df[ (df['age']>20) & (df['gender']=="female")]
print(df2,type())

```

# Query

```python
data={
    'name':['dipok','dipu','mahir','susmit','dipika','susmita','dipa'],
    'age':[21,20,23,24,20,23,22],
    'dept':['cse','swe','bba','agri','phy','chem','bmb'],
    'gender':['male','male','male','male','female','female','female',]
}
df=pd.DataFrame(data)
print(df)
print()
newdf=df.query("age>20 and gender== 'female'")
print(newdf)
```
