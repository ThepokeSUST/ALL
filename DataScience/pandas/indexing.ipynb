# Set_index 

```python
import pandas as pd

data={
    'name':['a','b','c'],
    'age':[1,2,3]
}


df=pd.DataFrame(data,)
print(df)
print()
df2=df.set_index('name')
print(df2)
print(df)
print(df.reset_index())
print(df)
print(df2.reset_index())
print(df2)

```
