# From another dataFrame 

```python
data={
    'a':[1,2,3,4],
    'b':[2,3,4,5]
}

df=pd.DataFrame(data)
print(df)
print(df['a'])
df2=pd.DataFrame(df[['a','b']])
print(df2)

series=pd.Series([1,2,3],index=[3,2,1],name='n')
print(series)
df=pd.DataFrame(series)
print(df)
```

```python
# Initialize data to Dicts of series.
d = {'one': pd.Series([10, 20, 30, 40],
                      index=['a', 'b', 'c', 'r'],name='dipu'),
     'two': pd.Series([10, 20, 30, 40],
                      index=['a', 'b', 'c', 'e'])}

# creates Dataframe.
df = pd.DataFrame(d)

print(df)
```

## from from_dict


```python
data = {0: {'Name': 'Alice', 'Age': 25}, 1: {'Name': 'Bob', 'Age': 30, 'dept':'swe'}}
df = pd.DataFrame.from_dict(data, orient='columns')
print(df)


data = {0: {'Name': 'Alice', 'Age': 25}, 1: {'Name': 'Bob', 'Age': 30}}
df = pd.DataFrame.from_dict(data, orient='index')
print(df)

df=pd.DataFrame(data)
print(df)
```

```python
data = [
    {'Name': 'Alice', 'Info': {'Age': 25, 'City': 'NY'}},
    {'Name': 'Bob', 'Info': {'Age': 30, 'City': 'LA'}}
]

df = pd.json_normalize(data)
print(df)
print(pd.DataFrame(data))
```
