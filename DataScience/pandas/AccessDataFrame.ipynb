# column

```python
import pandas as pd

data={
    'name':['dipok','dipu','alice'],
    'age':[21,23,'s']
}

df=pd.DataFrame(data)
print(df)
print()
col=df['age']
print(col)
print(type(col))
```

## Row and column

```python

data={
    'name':['dipok','dipu','alice'],
    'age':[21,23,'s']
}

df=pd.DataFrame(data)
print(df)

row=df.loc[2]
print(row,type(row))
row=df.iloc[2]
print(row,type(row))
```

## multiple column

```python

data={
    'name':['dipok','dipu','alice'],
    'age':[21,23,34],
    'dept':['cse','swe','eee']

}

df=pd.DataFrame(data)
print(df)
print()
col=df[['name']]
print(col,type(col))
col=df['name']
print()
print(col,type(col))
print()

col=df[['name','age']]
print(col,type(col))
```

## multiple row and column

```python
data={
    'name':['dipok','dipu','alice'],
    'age':[21,23,34],
    'dept':['cse','swe','eee']

}

df=pd.DataFrame(data,index=['a','b','c'])
print(df)
print()
row_col=df.loc['a':'b',['name','dept','age']]
print(row_col,type(row_col))
print()

row_col=df.loc['a':'b',:]
print(row_col)
rowcol=df.head(1)
print(rowcol,type(rowcol))
```

## condition

```python
data={
    'name':['dipok','dipu','alice'],
    'age':[21,23,34],
    'dept':['cse','swe','eee']

}
df=pd.DataFrame(data)
print(df)
print()
row=df[df['age']>21]
print(row,type(row))
print()

row=df[df['name']=='dipok']
print(row)
```

```python
data={
    'name':['dipok','dipu','alice'],
    'age':[21,23,34],
    'dept':['cse','swe','eee']

}
df=pd.DataFrame(data)
print(df)
print()

ele=df.at[0,'dept']
print(ele,type(ele))
```
